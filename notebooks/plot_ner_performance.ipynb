```python
import pandas as pd
import plotly.graph_objects as go
from plotly.colors import sample_colorscale
```

```python
COLOR_PALETTE = [
    "#ffffd9",
    "#edf8b1",
    "#c7e9b4",
    "#7fcdbb",
    "#41b6c4",
    "#1d91c0",
    "#225ea8",
    "#0c2c84",
]


def generate_palette(color_count: int) -> list[str]:
    """Sample a color scale for the requested number of colors.

    Returns
    -------
    list[str]
        A list of hex color codes sampled from the blue-to-purple color scale.
    """
    sample_points = [index / max(color_count - 1, 1) for index in range(color_count)]
    return sample_colorscale(COLOR_PALETTE, sample_points)


def apply_journal_theme(figure: go.Figure, y_axis_title: str) -> None:
    """Apply a clean, publication-ready layout to a Plotly figure."""
    figure.update_layout(
        template="simple_white",
        font={"family": "Arial, Helvetica, sans-serif", "size": 16, "color": "black"},
        plot_bgcolor="white",
        paper_bgcolor="white",
        margin={"l": 70, "r": 30, "t": 70, "b": 90},
    )
    figure.update_yaxes(
        title=y_axis_title.capitalize().replace("_", " "),
        range=[0, 1],
        tickfont={"size": 14},
        title_font={"size": 17},
        showgrid=True,
        gridcolor="rgba(0,0,0,0.08)",
        zeroline=False,
        ticks="outside",
    )
    figure.update_xaxes(
        tickfont={"size": 16},
        showgrid=False,
        tickangle=0,
        ticks="outside",
    )
```

## **Figure 1 — Effect of Prompt Strategy per Model**


```python
def plot_benchmark_llm_strategy(
    overall: pd.DataFrame, human_iaa: float, metric: str
) -> go.Figure:
    """Build the prompt-strategy comparison figure.

    Returns
    -------
    go.Figure
        A Plotly figure comparing performance of different models
        across prompt strategies.
    """
    data = overall[overall["scenario"] == "benchmark_strategies"].copy()

    normalization_rows = data[data["combo"] == "with_instructor_with_guidelines"].copy()
    normalization_rows["combo"] = "with_instructor_with_guidelines_normalization"
    normalization_rows[metric] = normalization_rows[f"{metric}_no_hallucination"]
    data = pd.concat([data, normalization_rows], ignore_index=True)

    combo_labels = {
        "no_instructor_no_guidelines": "No instructor<br>no guideline",
        "no_instructor_with_guidelines": "No instructor<br>with guideline",
        "with_instructor_no_guidelines": "With instructor<br>no guideline",
        "with_instructor_with_guidelines": "With instructor<br>with guideline",
        "with_instructor_with_guidelines_normalization": (
            "With instructor<br>with guideline and normalization"
        ),
    }
    combo_order = [combo for combo in combo_labels if combo in set(data["combo"])]
    max_per_combo = data.groupby("combo")[metric].apply(lambda s: s.round(2).max())

    models = data["model_safe"].unique()
    palette = generate_palette(len(models))

    figure = go.Figure()
    for model, color in zip(models, palette, strict=False):
        subset = (
            data[data["model_safe"] == model].set_index("combo").reindex(combo_order)
        )
        figure.add_bar(
            x=[combo_labels[combo] for combo in combo_order],
            y=subset[metric],
            name=model,
            marker={"color": color, "line": {"color": "white", "width": 1}},
            text=[
                f"<b>{v:.2f}</b>"
                if round(v, 2) == max_per_combo.get(combo)
                else f"{v:.2f}"
                for combo, v in subset[metric].items()
            ],
            textposition="outside",
            textfont={"size": 13},
        )

    figure.add_hline(
        y=human_iaa,
        line={"color": "#555555", "dash": "dash", "width": 1.5},
        annotation_text="Human IAA",
        annotation_position="right",
        annotation_font={"size": 14, "color": "#555555"},
    )

    apply_journal_theme(figure, metric)
    figure.update_layout(
        margin={"r": 100},
        hovermode=False,
        bargroupgap=0.08,
        legend={
            "orientation": "h",
            "yanchor": "bottom",
            "y": 1.02,
            "xanchor": "center",
            "x": 0.5,
            "font": {"size": 15},
        },
    )
    return figure
```

```python
strategy_figure = plot_benchmark_llm_strategy(overall, human_iaa=0.76, metric="f1")
strategy_figure.show()
```

```python
strategy_figure = plot_benchmark_llm_strategy(overall, human_iaa=0.76, metric="precision")
strategy_figure.show()
```

## Figure 2 — Standalone Models vs. Consensus


```python
# Set the sta
standalone_models = [
    "deepseek/deepseek-v4-pro",
    "google/gemma-4-31b-it",
    "openai/gpt-oss-120b",
    "qwen/qwen3.7-max",
]

# Set the metric to plot: f1_score, precision, or recall
METRIC = "precision"
```

```python
def plot_benchmark_consensus_strategy(
    overall: pd.DataFrame, standalone_models: list[str], metric: str
) -> go.Figure:
    """Build the standalone-models-vs-consensus figure.

    Returns
    -------
    go.Figure
        A Plotly figure comparing performance of standalone models
        against consensus models at different temperatures.
    """
    # Get the relevant data for standalone models
    solo = overall[
        (overall["scenario"] == "benchmark_models")
        & (overall["combo"] == "instructor_guidelines")
        & (overall["model_name"].isin(standalone_models))
    ]
    # Get the relevant data for consensus models
    cons = overall[
        (overall["scenario"] == "evaluation") & (overall["combo"] == "consensus")
    ].set_index("model_safe")
    # Plot the data
    labels = [*solo["model_name"], "Consensus<br>(T=1)", "Consensus<br>(T=1, 2)"]
    scores = [
        *solo[metric],
        cons.loc["temp_1", metric],
        cons.loc["temp_1_and_2", metric],
    ]
    figure = go.Figure(
        go.Bar(
            x=labels,
            y=scores,
            marker_color=generate_palette(len(labels)),
            marker_line={"color": "white", "width": 1},
            text=[f"{s:.2f}" for s in scores],
            textposition="outside",
            textfont_size=13,
        )
    )
    apply_journal_theme(figure, metric)
    figure.update_xaxes(
        tickangle=-10,
        tickson="labels",  # ou centrage standard
        ticklabelposition="outside",
    )
    figure.update_layout(hovermode=False, showlegend=False, bargap=0.3)
    return figure
```

```python
consensus_figure = plot_benchmark_consensus_strategy(
    overall,
    standalone_models=[
        "deepseek/deepseek-v4-pro",
        "google/gemma-4-31b-it",
        "openai/gpt-oss-120b",
        "qwen/qwen3.7-max",
    ],
    metric="precision",
)
consensus_figure.show()
```

```python
def get_base_family(name: str) -> str:
    """Normalize model display or checkpoint name.

    Returns
    -------
    str
        The base architecture family name, with any fine-tuning suffix removed.
    """
    clean_name = name.split("/")[-1].lower()
    return clean_name.replace("-finetuned", "").replace("_finetuned", "")


def get_generation_group(family_name: str) -> str:
    """Map architecture family name to its broad GLiNER generation group.

    Returns
    -------
    str
        One of 'GLiNER-BioMed', 'GLiNER2', or 'GLiNER2.5'.
    """
    if "biomed" in family_name:
        return "GLiNER-BioMed"
    if "gliner2.5" in family_name:
        return "GLiNER2.5"
    return "GLiNER2"


def plot_benchmark_gliner_models(
    df: pd.DataFrame,
    human_iaa: float,
    metric: str = "F1",
    category: str = "OVERALL_MICRO",
) -> go.Figure:
    """Build GLiNER comparison figure.

    Returns
    -------
    go.Figure
        A Plotly figure comparing performance of different GLiNER models
        with right-angled bracket connectors, centered delta annotations,
        and top legend for major model generations.
    """
    data = df[df["category"] == category].copy()

    # Identify base model family, generation group, and fine-tuning flag
    data["base_family"] = data["Name"].apply(get_base_family)
    data["generation_group"] = data["base_family"].apply(get_generation_group)
    data["is_finetuned"] = data["Name"].str.contains("finetuned", case=False, na=False)
    data["display_label"] = data["Name"].apply(lambda s: s.split("/")[-1])

    # Assign distinct color per architecture family
    unique_families = list(dict.fromkeys(data["base_family"]))
    palette = generate_palette(len(unique_families))
    family_to_color = dict(zip(unique_families, palette, strict=False))

    # Sort to place base models and fine-tuned counterparts side-by-side
    data["family_order"] = data["base_family"].map(
        {fam: i for i, fam in enumerate(unique_families)}
    )
    data = data.sort_values(
        by=["family_order", "is_finetuned"], ascending=[True, True]
    ).reset_index(drop=True)
    # Format x-axis tick labels to bold the best performing model
    data["x_axis_label"] = data["display_label"]

    figure = go.Figure()

    # 1. Render bars first with categorical x labels
    for _, row in data.iterrows():
        is_ft = row["is_finetuned"]
        val = row[metric]
        color = family_to_color[row["base_family"]]
        x_label = row["x_axis_label"]

        figure.add_bar(
            x=[x_label],
            y=[val],
            name=row["Name"],
            marker={
                "color": color,
                "line": {"color": "white", "width": 1.2},
                "pattern": {
                    "shape": "/" if is_ft else "",
                    "solidity": 0.35,
                    "fgcolor": "white",
                },
            },
            text=[f"{val:.2f}"],
            textposition="outside",
            textfont={"size": 13},
            showlegend=False,
        )

    # 2. Add legend scatter traces with explicit category binding
    generation_order = ["GLiNER-BioMed", "GLiNER2", "GLiNER2.5"]
    first_category_x = data["x_axis_label"].iloc[0]
    for group in generation_order:
        fam_matches = data[data["generation_group"] == group]
        if not fam_matches.empty:
            rep_family = fam_matches.iloc[0]["base_family"]
            rep_color = family_to_color[rep_family]
            figure.add_trace(
                go.Scatter(
                    x=[first_category_x],
                    y=[None],
                    mode="markers",
                    name=group,
                    marker={"size": 13, "symbol": "square", "color": rep_color},
                    showlegend=True,
                )
            )

    # 3. Draw right-angled connector lines and centered delta annotations
    for family in unique_families:
        fam_subset = data[data["base_family"] == family]
        base_rows = fam_subset[~fam_subset["is_finetuned"]]
        ft_rows = fam_subset[fam_subset["is_finetuned"]]

        if not base_rows.empty and not ft_rows.empty:
            base_row = base_rows.iloc[0]
            ft_row = ft_rows.iloc[0]

            x_base = base_row["x_axis_label"]
            x_ft = ft_row["x_axis_label"]

            y_base = base_row[metric]
            y_ft = ft_row[metric]
            delta = y_ft - y_base

            # Bridge height above the tallest bar in the pair
            h_bridge = max(y_base, y_ft) + 0.1

            # Single square connector path using direct category labels
            path = (
                f"M {x_base},{y_base} "
                f"L {x_base},{h_bridge} "
                f"L {x_ft},{h_bridge} "
                f"L {x_ft},{y_ft}"
            )
            figure.add_shape(
                type="path",
                path=path,
                fillcolor="rgba(0,0,0,0)",
                line={"color": "#111111", "width": 1.2},
            )

            # Center delta annotation above the horizontal segment
            delta_sign = "+" if delta >= 0 else ""
            figure.add_annotation(
                x=x_ft,
                y=h_bridge + 0.05,
                xshift=-40,
                text=f"<b>{delta_sign}{delta:.2f}</b>",
                showarrow=False,
                font={"size": 13, "color": "#111111"},
                align="center",
            )

    # 4. Add Human IAA reference baseline
    figure.add_hline(
        y=human_iaa,
        line={"color": "#555555", "dash": "dash", "width": 1.5},
        annotation_text="Human IAA",
        annotation_position="right",
        annotation_font={"size": 14, "color": "#555555"},
    )

    apply_journal_theme(figure, metric)
    figure.update_layout(
        hovermode=False,
        margin={"l": 80, "r": 140, "t": 90, "b": 150},
        xaxis={
            "type": "category",
            "tickangle": -25,
        },
        legend={
            "orientation": "h",
            "yanchor": "bottom",
            "y": 1.02,
            "xanchor": "center",
            "x": 0.5,
            "font": {"size": 15},
        },
    )

    return figure
```

```python
fig = plot_benchmark_gliner_models(
    df=df,
    human_iaa=0.76,
    metric="F1",
    category="OVERALL_MICRO",
)
fig.show()
fig.write_image("../plots/gliner_benchmark_f1.png", scale=10, width=1200, height=600)
```
